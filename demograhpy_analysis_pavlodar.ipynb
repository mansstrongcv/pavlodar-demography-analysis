```python
import pandas as pd


!pip install -q xlrd


url_population = "https://github.com/mansstrongcv/pavlodar-demography-analysis/raw/ec595228af572f082ed7e44f6e2aa69e236fc744/data/%D0%A7%D0%B8%D1%81%D0%BB%D0%B5%D0%BD%D0%BD%D0%BE%D1%81%D1%82%D1%8C%20%D0%BD%D0%B0%D1%81%D0%B5%D0%BB%D0%B5%D0%BD%D0%B8%D1%8F%20(3).xls"
url_migration = "https://github.com/mansstrongcv/pavlodar-demography-analysis/raw/ec595228af572f082ed7e44f6e2aa69e236fc744/data/%D0%92%D0%BD%D0%B5%D1%88%D0%BD%D1%8F%D1%8F%20%D0%BC%D0%B8%D0%B3%D1%80%D0%B0%D1%86%D0%B8%D1%8F%20%D0%BF%D0%BE%20%D1%81%D1%82%D1%80%D0%B0%D0%BD%D0%B0%D0%BC.xls"
url_growth = "https://github.com/mansstrongcv/pavlodar-demography-analysis/raw/ec595228af572f082ed7e44f6e2aa69e236fc744/data/%D0%95%D1%81%D1%82%D0%B5%D1%81%D1%82%D0%B2%D0%B5%D0%BD%D0%BD%D1%8B%D0%B9%20%D0%BF%D1%80%D0%B8%D1%80%D0%BE%D1%81%D1%82.xls"


def clean_df(url, header_row=3):
    temp_df = pd.read_excel(url, header=header_row)
    cleaned = temp_df.T
    cleaned.columns = cleaned.iloc[0]
    cleaned = cleaned[1:]
    cleaned.index.name = 'Год'


    cleaned = cleaned.loc[:, cleaned.columns.notna()]
    return cleaned


df_pop_final = clean_df(url_population, header_row=3)
df_migration_final = clean_df(url_migration, header_row=3)

df_growth_final = clean_df(url_growth, header_row=5)


print("Чистая таблица населения:")
display(df_pop_final.head())

print("\nЧистая таблица миграции:")
display(df_migration_final.head())

print("\nЧистая таблица прироста:")
display(df_growth_final.head())
```

```python

df_growth_raw = pd.read_excel(url_growth, header=4)


df_growth_final = df_growth_raw.T


df_growth_final.columns = df_growth_final.iloc[0]


df_growth_final = df_growth_final[1:]
df_growth_final.index.name = 'Год'
df_growth_final = df_growth_final.iloc[1:]

display(df_growth_final.head())
```

```python

df_growth_raw = pd.read_excel(url_growth, header=None)


years = df_growth_raw.iloc[2, 1:].values


data_rows = df_growth_raw.iloc[3:]


df_growth_final = data_rows.iloc[:, 1:].T


df_growth_final.columns = data_rows.iloc[:, 0].values


df_growth_final.index = years
df_growth_final.index.name = 'Год'


df_growth_final = df_growth_final.loc[:, df_growth_final.columns.notna()]


df_growth_final.index = df_growth_final.index.astype(int)


display(df_growth_final.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


df_pop_renamed = df_pop_final.add_prefix('Население_')
df_mig_renamed = df_migration_final.add_prefix('Миграция_')
df_growth_renamed = df_growth_final.add_prefix('Прирост_')


df_combined = df_pop_renamed.join(df_mig_renamed, how='outer').join(df_growth_renamed, how='outer')


df_combined = df_combined.sort_index()

print("Объединенный датафрейм (первые строки):")
display(df_combined.head())
print(f"Доступный диапазон лет: с {df_combined.index.min()} по {df_combined.index.max()}")
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_theme(style="whitegrid")
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 12), sharex=False, gridspec_kw={'height_ratios': [1, 1.2]})


pop_cols = [c for c in df_combined.columns if 'Население_' in c and ('область' in c.lower() or 'павлодарская' in c.lower())]


selected_pop = df_combined[pop_cols[0]]
if isinstance(selected_pop, pd.DataFrame):
    data_pop = selected_pop.iloc[:, 0].dropna()
else:
    data_pop = selected_pop.dropna()


data_pop = data_pop.astype(float)

sns.lineplot(x=data_pop.index, y=data_pop.values, ax=ax1, color='#1f77b4', marker='o', linewidth=2.5)
ax1.set_title('Динамика численности населения Павлодарской области (тыс. человек)', fontsize=14, fontweight='bold', pad=15)
ax1.set_ylabel('тыс. человек', fontsize=12)
ax1.set_xlabel('Год', fontsize=12)
ax1.set_xlim(data_pop.index.min() - 1, data_pop.index.max() + 1)

for x, y in zip(data_pop.index, data_pop.values):
    ax1.annotate(f'{y:.1f}', (x, y), textcoords="offset points", xytext=(0,10), ha='center', fontsize=9, fontweight='semibold')


mig_col = 'Миграция_Всего'
growth_cols = [c for c in df_combined.columns if 'Прирост_' in c and ('область' in c.lower() or 'павлодарская' in c.lower())]


selected_growth = df_combined[growth_cols[0]]
if isinstance(selected_growth, pd.DataFrame):
    data_growth = selected_growth.iloc[:, 0].dropna()
else:
    data_growth = selected_growth.dropna()

selected_mig = df_combined[mig_col]
if isinstance(selected_mig, pd.DataFrame):
    data_mig = selected_mig.iloc[:, 0].dropna()
else:
    data_mig = selected_mig.dropna()


common_years = data_mig.index.intersection(data_growth.index)
data_mig_filtered = data_mig.loc[common_years].astype(float)
data_growth_filtered = data_growth.loc[common_years].astype(float)

x_indices = np.array(common_years)
width = 0.35

rects1 = ax2.bar(x_indices - width/2, data_growth_filtered.values, width, label='Естественный прирост', color='#2ca02c', alpha=0.85)
rects2 = ax2.bar(x_indices + width/2, data_mig_filtered.values, width, label='Внешняя миграция (сальдо)', color='#d62728', alpha=0.85)

ax2.set_title('Компоненты изменения численности населения (человек)', fontsize=14, fontweight='bold', pad=15)
ax2.set_ylabel('Количество человек', fontsize=12)
ax2.set_xlabel('Год', fontsize=12)
ax2.set_xticks(x_indices)
ax2.set_xticklabels(x_indices, rotation=45)
ax2.axhline(0, color='black', linewidth=1, linestyle='--')
ax2.legend(fontsize=11, loc='upper left')

# Авто-подписи значений на столбцах
def autolabel(rects, ax):
    for rect in rects:
        height = rect.get_height()
        if not np.isnan(height):
            va_dir = 'bottom' if height >= 0 else 'top'
            xy_text = (0, 3) if height >= 0 else (0, -12)
            ax.annotate(f'{int(height)}',
                        xy=(rect.get_x() + rect.get_width() / 2, height),
                        xytext=xy_text,
                        textcoords="offset points",
                        ha='center', va=va_dir, fontsize=8, fontweight='semibold')

autolabel(rects1, ax2)
autolabel(rects2, ax2)

plt.tight_layout()
plt.show()
```

```python
# Находим подходящие колонки с учетом добавленных ранее префиксов
growth_cols = [c for c in df_combined.columns if 'Прирост_' in c and ('область' in c.lower() or 'павлодарская' in c.lower())]
mig_cols = [c for c in df_combined.columns if 'Миграция_Всего' in c]

# Безопасно извлекаем данные как одномерные Series (на случай дублирования колонок)
if growth_cols:
    selected_growth = df_combined[growth_cols[0]]
    series_growth = selected_growth.iloc[:, 0] if isinstance(selected_growth, pd.DataFrame) else selected_growth
    total_growth = pd.to_numeric(series_growth, errors='coerce').sum()
else:
    total_growth = 0

if mig_cols:
    selected_mig = df_combined[mig_cols[0]]
    series_mig = selected_mig.iloc[:, 0] if isinstance(selected_mig, pd.DataFrame) else selected_mig
    total_migration = pd.to_numeric(series_mig, errors='coerce').sum()
else:
    total_migration = 0

print(f"Суммарный естественный прирост (Павлодарская область): {int(total_growth)} чел.")
print(f"Суммарное сальдо внешней миграции: {int(total_migration)} чел.")
```
